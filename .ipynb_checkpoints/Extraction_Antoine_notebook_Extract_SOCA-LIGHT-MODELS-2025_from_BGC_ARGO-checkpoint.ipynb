## 5) Extraction of SOCA-light-2025 profiles

### Extract light profiles from input file and BGC-Argo file
***

> * The path of the BGC-Argo file needs to be specified in the input file.
> * The output files will be saved in the corresponding output folder for each variable.
> * Users will get outputs in the corresponding WMO folder. **No need to change anything in this NOTEBOOK.**

For **SOCA Light 2025**, each light variable (PAR, ED380, ED412, and ED490) provides **two output variables**: **one at the local time of the float** and **other at local noon**. 


For more details, please refer to the Quality Information Document: 
[Copernicus Marine MULTIOBS Product Description](https://data.marine.copernicus.eu/product/MULTIOBS_GLO_BIO_BGC_3D_REP_015_010/description)

*******

```python

# --- 3. PROCESSING LOOP ---
# for i in tqdm(range(53000, len(Inputs)), desc="Processing Argo Profiles"):
for i in tqdm(range(len(Inputs)), desc="Processing Argo Profiles"):
    try:
        # 1. Determine if it's SD or SR
        prefix = "SD" if "SD" in str(Inputs.BGC_ARGO_NC_FILE[i]) else "SR"
        
        # 2. Build the path
        full_nc_path = os.path.join(
            BASE_DAC_PATH,
            str(Inputs.DAC[i]),
            str(Inputs.WMO[i]),
            "profiles",
            f"{prefix}{Inputs.WMO[i]}_{Inputs.N_CYCLE[i]}.nc"
        )
        
        # 3. Numeric Conversion for Satellite Inputs (Prevents str/str division error)
        try:
            RRS412 = float(Inputs.RHO412[i])/np.pi
            RRS443 = float(Inputs.RHO443[i])/np.pi
            RRS490 = float(Inputs.RHO490[i])/np.pi
            RRS555 = float(Inputs.RHO555[i])/np.pi
            RRS670 = float(Inputs.RHO670[i])/np.pi
            PAR = float(Inputs.PAR[i])
            KD490 = float(Inputs.KD490[i])
        except (ValueError, TypeError):
            # Skip profile if satellite inputs are corrupt strings
            continue

        base_dir = Path(OUT_PATH) 
        # filename = f"{Inputs.WMO[i]}_{Inputs.N_CYCLE[i]}_SOCA_LIGHT.txt"
        # output_file = str(base_dir / filename)

        # 2. Define the subfolder name (using the actual WMO value from your input)
        wmo_folder = str(Inputs.WMO[i])
        filename = f"{Inputs.WMO[i]}_{Inputs.N_CYCLE[i]}_SOCA_LIGHT.txt"

        # 3. Create the full path to the folder
        target_dir = base_dir / wmo_folder

        # 4. Create the folder if it doesn't exist
        # parents=True creates any missing base folders; exist_ok=True prevents errors if it's already there
        target_dir.mkdir(parents=True, exist_ok=True)

        # 5. Define the final file path
        output_file = target_dir / filename        
        
        # Initialize output variables as NaN
        PAR_ADJUSTED = PAR_ADJUSTED_noon = np.nan
        ED380_ADJUSTED = ED380_ADJUSTED_noon = np.nan
        ED412_ADJUSTED = ED412_ADJUSTED_noon = np.nan
        ED490_ADJUSTED = ED490_ADJUSTED_noon = np.nan
        depth_output = np.linspace(0, 250, 51)

        if os.path.isfile(full_nc_path):
            NC = NetCDFFile(str(full_nc_path))

            # Extract Parameter/Mode
            PARAMETER = NC.variables['PARAMETER']
            PARAMETER = ''.join(str(s, encoding='UTF-8') for s in PARAMETER).split()
            DATA_M = NC.variables['PARAMETER_DATA_MODE']
            DATA_M = ''.join(str(s, encoding='UTF-8') for s in DATA_M)
            
            TEMP_MODE = DATA_M[PARAMETER.index('TEMP')]
            PSAL_MODE = DATA_M[PARAMETER.index('PSAL')]

            if (TEMP_MODE == 'R') and (PSAL_MODE == 'R'):
                p_var, t_var, s_var = 'PRES', 'TEMP', 'PSAL'
                pq_var, tq_var, sq_var = 'PRES_QC', 'TEMP_QC', 'PSAL_QC'
                use_data = True
            elif (TEMP_MODE in ['D', 'A']) and (PSAL_MODE in ['D', 'A']):
                p_var, t_var, s_var = 'PRES_ADJUSTED', 'TEMP_ADJUSTED', 'PSAL_ADJUSTED'
                pq_var, tq_var, sq_var = 'PRES_ADJUSTED_QC', 'TEMP_ADJUSTED_QC', 'PSAL_ADJUSTED_QC'
                use_data = True
            else:
                use_data = False

            if use_data:
                pres_raw = NC.variables[p_var][:]
                temp_raw = NC.variables[t_var][:]
                sal_raw = NC.variables[s_var][:]              
                pres_qc_raw = NC.variables[pq_var][:]
                temp_qc_raw = NC.variables[tq_var][:]
                sal_qc_raw = NC.variables[sq_var][:]
                
                LAT = float(NC.variables['LATITUDE'][:])
                LON = float(NC.variables['LONGITUDE'][:])

                pres1, temp1, sal1 = pres_raw.filled(np.nan), temp_raw.filled(np.nan), sal_raw.filled(np.nan)
                pres_qc, temp_qc, sal_qc = pres_qc_raw.filled(np.nan), temp_qc_raw.filled(np.nan), sal_qc_raw.filled(np.nan)

                pres1[(pres_qc == b'3') | (pres_qc == b'4')] = np.nan
                temp1[(temp_qc == b'3') | (temp_qc == b'4')] = np.nan
                sal1[(sal_qc == b'3') | (sal_qc == b'4')] = np.nan

                mask = (~np.isnan(pres1)) & (~np.isnan(sal1)) & (~np.isnan(temp1))
                pres2, temp2, sal2 = pres1[mask], temp1[mask], sal1[mask]

                if (len(pres2[pres2 <= 250]) >= 15) and (len(pres2[pres2 <= 50]) >= 5):
                    # --- DENSITY & MLD ---
                    SA = gsw.SA_from_SP(sal2, pres2, LON, LAT)
                    CT = gsw.CT_from_t(SA, temp2, pres2)
                    dens = gsw.density.sigma0(SA, CT)
                    
                    # Robust MLD Guard
                    MLD_1 = np.nan
                    if len(pres2) > 5:
                        try:
                            MLD_1 = find_MLD(Z=pres2, dens=dens, threshold=0.03)
                        except (ValueError, IndexError):
                            MLD_1 = np.nan

                    # --- DATE & TIME ---
                    JULD = NC.variables['JULD'][:]
                    date_prof = pd.to_datetime('19500101', format='%Y%m%d') + timedelta(days=float(JULD))
                    date_str = date_prof.strftime('%Y%m%d')
                    DOY = int(date_prof.strftime('%j'))
                    sin_doy, cos_doy = np.sin(rad_date(DOY)), np.cos(rad_date(DOY))

                    # --- SOLAR ELEVATION ---
                    timezone_str = tf.certain_timezone_at(lat=LAT, lng=LON)
                    timezone = pytz.timezone(timezone_str)
                    LC = date_prof + timezone.utcoffset(date_prof)
                    LC_Hour_Deci = LC.hour + (LC.minute / 60.0)

                    solar_elevation = calculate_solar_elevation(date_str, LC_Hour_Deci, LAT, LON)  
                    solar_elevation_noon = calculate_solar_elevation(date_str, 12, LAT, LON)  

                    # --- LIGHT MODEL PROCESSING WITH < 5 DEGREE GUARD ---
                    if not np.isnan(MLD_1):
                        common_args = {
                            "PAR_SAT": PAR, "MLD": np.log10(MLD_1), "ZPD_MOREL": derive_ZPD(KD490),
                            "RRS412": RRS412, "RRS443": RRS443, "RRS490": RRS490, "RRS555": RRS555, "RRS670": RRS670,
                            "sin_doy": sin_doy, "cos_doy": cos_doy, "temp_cmems": temp2, "sal_cmems": sal2,
                            "depths_cmems": pres2, "depth_output": depth_output
                        }

                        # Current Time Calculation (Guard: Elevation >= 5)
                        if solar_elevation >= 5:
                            log10_sol = np.log10(solar_elevation)
                            par_inputs = INPUTS_PAR_ADJUSTED_2025(solar_elevation_pvlib=log10_sol, **common_args)
                            PAR_ADJUSTED, _ = SOCA_PAR_ADJUSTED_2025(par_inputs, depth_output)
                            
                            ed380_in = INPUTS_ED380_ADJUSTED_2025(solar_elevation_pvlib=log10_sol, **common_args)
                            ED380_ADJUSTED, _ = SOCA_ED380_ADJUSTED_2025(ed380_in, depth_output)

                            ed412_in = INPUTS_ED412_ADJUSTED_2025(solar_elevation_pvlib=log10_sol, **common_args)
                            ED412_ADJUSTED, _ = SOCA_ED412_ADJUSTED_2025(ed412_in, depth_output)

                            ed490_in = INPUTS_ED490_ADJUSTED_2025(solar_elevation_pvlib=log10_sol, **common_args)
                            ED490_ADJUSTED, _ = SOCA_ED490_ADJUSTED_2025(ed490_in, depth_output)

                        # Noon Calculation (Guard: Elevation >= 5)
                        if solar_elevation_noon >= 5:
                            log10_noon = np.log10(solar_elevation_noon)
                            par_inputs_n = INPUTS_PAR_ADJUSTED_2025(solar_elevation_pvlib=log10_noon, **common_args)
                            PAR_ADJUSTED_noon, _ = SOCA_PAR_ADJUSTED_2025(par_inputs_n, depth_output)
                            
                            ed380_in_n = INPUTS_ED380_ADJUSTED_2025(solar_elevation_pvlib=log10_noon, **common_args)
                            ED380_ADJUSTED_noon, _ = SOCA_ED380_ADJUSTED_2025(ed380_in_n, depth_output)

                            ed412_in_n = INPUTS_ED412_ADJUSTED_2025(solar_elevation_pvlib=log10_noon, **common_args)
                            ED412_ADJUSTED_noon, _ = SOCA_ED412_ADJUSTED_2025(ed412_in_n, depth_output)

                            ed490_in_n = INPUTS_ED490_ADJUSTED_2025(solar_elevation_pvlib=log10_noon, **common_args)
                            ED490_ADJUSTED_noon, _ = SOCA_ED490_ADJUSTED_2025(ed490_in_n, depth_output)

            NC.close()

        # --- SAVE TO CSV ---
        db_soca_light = {
            'Depth': depth_output, 
            'PAR_ADJUSTED': PAR_ADJUSTED, 'PAR_ADJUSTED_NOON': PAR_ADJUSTED_noon,
            'ED380_ADJUSTED': ED380_ADJUSTED, 'ED380_ADJUSTED_NOON': ED380_ADJUSTED_noon,
            'ED412_ADJUSTED': ED412_ADJUSTED, 'ED412_ADJUSTED_NOON': ED412_ADJUSTED_noon,
            'ED490_ADJUSTED': ED490_ADJUSTED, 'ED490_ADJUSTED_NOON': ED490_ADJUSTED_noon
        } 
        
        db_soca_light = {k: (np.round(v, 5) if k != 'Depth' else v) for k, v in db_soca_light.items()}
        pd.DataFrame(db_soca_light).to_csv(output_file, sep=',', index=False)

    except Exception as e:
        print(f"Error at index {i}: {e}")
        continue
```
